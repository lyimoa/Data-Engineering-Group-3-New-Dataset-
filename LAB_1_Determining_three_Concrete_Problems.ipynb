## LAB 1 - Determining three Concrete Problems

### Step 1: Import and Set-up

```python
#!pip install kagglehub -q

import os
import kagglehub
import pandas as pd

# Download latest version of the dataset
path = kagglehub.dataset_download("prasad22/healthcare-dataset")
print("Path to dataset files:", path)

# List files in the downloaded dataset folder
print(os.listdir(path))
```

### Step 1a: Loading the dataset into a DataFrame to inspect shape/columns

```python
df = pd.read_csv(os.path.join(path, "healthcare_dataset.csv"))

print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nFirst 5 rows:")
df.head()
```

### Step 2: Basic data quality scan

### Step 2a: Checking Missing values per column

```python
print("=== Missing values per column ===")
print(df.isnull().sum())
```

### Step 2b: Checking Dublicate rows

```python
print("\n=== Duplicate rows ===")
print("Full duplicate rows:", df.duplicated().sum())
```

### Step 2c: Checking Quick stats on numeric columns

```python
print("\n=== Quick stats on numeric columns ===")
df.describe()
```

### Step 2d: Checking Data types

```python
print("\n=== Data types ===")
print(df.dtypes)
```

### Step 2e: Checking Text Consistency

```python
print("=== Sample of Name, Doctor, Hospital columns (check capitalization) ===")
print(df[['Name', 'Doctor', 'Hospital']].head(10))

print("\n=== Unique values in categorical columns ===")
for col in ['Gender', 'Blood Type', 'Medical Condition', 'Insurance Provider', 'Admission Type', 'Test Results']:
    print(f"\n{col}: {df[col].unique()}")
```

### Step 2f: Checking Date Logic

```python
print("\n=== Convert dates and check for discharge before admission ===")
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'])

bad_dates = df[df['Discharge Date'] < df['Date of Admission']]
print("Rows where discharge is before admission:", len(bad_dates))
```
